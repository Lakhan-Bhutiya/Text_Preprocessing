```python
# Importing required NLP libraries
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download essential NLP datasets/models
nltk.download('punkt')      # Tokenizer model
nltk.download('stopwords')  # Stopwords list
nltk.download('wordnet')    # Lemmatizer dictionary
```

```python
text = """

Artificial intelligence (AI) is the capability of computational systems to perform tasks typically associated with human intelligence, such as learning, reasoning, problem-solving, perception, and decision-making. It is a field of research in computer science that develops and studies methods and software that enable machines to perceive their environment and use learning and intelligence to take actions that maximize their chances of achieving defined goals.
"""
```

```python
# Step 1: Convert all text to lowercase for consistency
clean_text = text.lower()

# Step 2: Remove punctuation before tokenization
# This avoids tokens like ',' '.' '/' appearing
clean_text = re.sub(r"[^\w\s]", " ", clean_text)

# Step 3: Tokenize words and sentences
from nltk.tokenize import word_tokenize, sent_tokenize
tokens = word_tokenize(clean_text)      # Word-level tokenization
sentences = sent_tokenize(clean_text)   # Sentence-level tokenization

tokens[:10], sentences[:]   # Show sample
```

```python
# Stopwords are common words like "the", "and", "is" that add no meaning
stop_words = set(stopwords.words("english"))

# Remove stopwords from tokens
tokens_no_stop = [w for w in tokens if w not in stop_words]

tokens_no_stop[:10]
```

```python
# Stemming reduces words to root form by chopping ends
# Example: "running" → "run", "studies" → "studi"

p = PorterStemmer()
stemmed_tokens = [p.stem(w) for w in tokens_no_stop]

stemmed_tokens[:10]
```

```python
# Lemmatization reduces words to proper dictionary form
# Example: "better" → "good", "languages" → "language"

lem = WordNetLemmatizer()
lemmatized_tokens = [lem.lemmatize(w) for w in tokens_no_stop]

lemmatized_tokens[:10]
```

```python
# POS (Part-Of-Speech) tagging identifies nouns, verbs, adjectives, etc.
# Example: "Emma" → NNP (Proper noun)

nltk.download('averaged_perceptron_tagger')
pos_tags = nltk.pos_tag(tokens)

pos_tags[:10]
```

```python
# NE Chunking identifies real-world entities like PEOPLE, PLACES, ORGANIZATIONS

nltk.download('maxent_ne_chunker')
nltk.download('words')

chunked = nltk.ne_chunk(pos_tags)
print(chunked[:5])      # Prints named entity tree
```

```python
# N-grams represent word sequences and capture context
# bigrams = 2-word pairs, trigrams = 3-word groups

from nltk.util import bigrams, trigrams, ngrams

bigrams_list = list(bigrams(tokens))
trigrams_list = list(trigrams(tokens))
sixgrams_list = list(ngrams(tokens, 6))

bigrams_list[:5], trigrams_list[:5], sixgrams_list[:3]
```

```python
# Creating a clean corpus where each sentence is cleaned separately
# Useful for ML models like TF-IDF later

clean_corpus = []
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()

for sent in sentences:
    words = word_tokenize(sent)
    cleaned = []
    
    for w in words:
        if w not in stop_words:                 # Remove stopwords
            w = stemmer.stem(w)                 # Stem words
            w = re.sub(r'[^\w\s]', '', w)       # Remove punctuation
            if w.strip() != "":
                cleaned.append(w)
    
    clean_corpus.append(" ".join(cleaned))

clean_corpus[:5]
```

```python
# Frequency distribution shows most common words in your cleaned text
# Useful for analysis and understanding key terms

from nltk.probability import FreqDist

fd = FreqDist(tokens_no_stop)
fd.most_common(10)
```
